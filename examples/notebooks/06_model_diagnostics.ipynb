# Model Diagnostics Tutorial

This notebook demonstrates how to run the comprehensive diagnostic checks
implemented in `industrialstats.analysis.diagnostics.ModelDiagnostics`.

```python
import math
import numpy as np
import pandas as pd
import statsmodels.api as sm

from industrialstats.analysis.diagnostics import ModelDiagnostics

rng = np.random.default_rng(42)
x1 = rng.normal(size=160)
x2 = rng.normal(size=160)
noise = rng.normal(scale=0.5, size=160)
y = 1.5 + 2.1 * x1 - 1.4 * x2 + noise
data = pd.DataFrame({"y": y, "x1": x1, "x2": x2})
model = sm.OLS(data["y"], sm.add_constant(data[["x1", "x2"]])).fit()
model_result = {
    "model_object": model,
    "residuals": model.resid,
    "fitted_values": model.fittedvalues,
    "model_metrics": {
        "R2": model.rsquared,
        "RMSE": math.sqrt(model.mse_resid),
    },
}
diagnostics = ModelDiagnostics(model_result, data)
tests = diagnostics.assumption_tests()
tests
```

```python
adequacy = diagnostics.model_adequacy()
recommendations = diagnostics.recommendation_system()
adequacy['assumptions'], recommendations
```
