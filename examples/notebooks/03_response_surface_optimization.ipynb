# Response Surface Optimisation Tutorial
This notebook demonstrates steepest ascent, ridge analysis, and multi-response optimisation for a quadratic model following Box & Draper (2007).

```python
import numpy as np
from industrialstats.designs.base import Factor
from industrialstats.designs.response_surface import ResponseSurfaceDesign

factors = [Factor('x1', [0, 1], factor_type='continuous'), Factor('x2', [0, 1], factor_type='continuous')]
design = ResponseSurfaceDesign(factors, design_type='CCD', center_points=3)
matrix = design.generate_design()
coded = design._get_design_matrix_coded()
coefficients = {
    'Intercept': 15.0,
    'x1': 1.2,
    'x2': -0.8,
    'x1²': 0.6,
    'x2²': 0.4,
    'x1*x2': 0.2,
}
quad = np.array([[coefficients['x1²'], coefficients['x1*x2'] / 2], [coefficients['x1*x2'] / 2, coefficients['x2²']]])
linear = np.array([coefficients['x1'], coefficients['x2']])
response = coefficients['Intercept'] + coded @ linear + np.sum((coded @ quad) * coded, axis=1)
results = design.response_surface_analysis(response.tolist())
steepest = design.steepest_ascent(results, n_steps=5)
ridge = design.ridge_analysis(results, radii=[0.5, 1.0])
multi = design.multiple_response_optimization({'yield': results})
steepest['path'].head()
```
