# Introduction to Design of Experiments

This beginner-friendly notebook walks through a simple $2^2$ factorial experiment and highlights key principles of DOE.

## Basic Concepts

- **Factors**: controllable variables in the experiment.
- **Levels**: settings for each factor (e.g., low and high).
- **Responses**: measured outcomes affected by factors.
- **Interactions**: combined effects of factors acting together.

```python
import pandas as pd
import plotly.express as px
import ipywidgets as widgets
from industrialstats.designs.base import Factor
from industrialstats.designs.factorial import FactorialDesign
from industrialstats.utils.data_generation import DataSimulator
```

```python
# build a 2^2 factorial design
factors = [Factor("A", [-1, 1]), Factor("B", [-1, 1])]
design = FactorialDesign(factors, randomize=False)
design_matrix = design.generate_design()
design_matrix
```

```python
def explore(a=1.0, b=1.0, interaction=0.0, noise=1.0):
    sim = DataSimulator(seed=0)
    responses = sim.simulate_factorial_response(
        design_matrix,
        main_effects={"A": a, "B": b},
        interactions={("A", "B"): interaction},
        noise_level=noise,
    )
    df = design_matrix.copy()
    df["Response"] = responses
    display(df)
    fig = px.scatter(df, x="A", y="B", color="Response", size="Response",
                     title="Simulated Responses")
    fig.show()

widgets.interact(
    explore,
    a=(-2.0, 2.0, 0.5),
    b=(-2.0, 2.0, 0.5),
    interaction=(-2.0, 2.0, 0.5),
    noise=(0.0, 2.0, 0.1),
)
```

## Interpreting Results

- Compare response averages at each factor level to assess main effects.
- Examine how the effect of one factor changes across levels of the other to detect interactions.
- Look for the combination of factor levels that maximizes or minimizes the response.

## Common Pitfalls

- Changing more than one factor at a time when exploring effects.
- Ignoring interactions, which can mask important findings.
- Forgetting to randomize run order, leading to biased results.
